## 1. Missing Values


### Step 1 – Understand the Concept
Own Words; Missing values mean some cells in the dataset do not contain useful data.

Definition: A missing value is a value that is absent or unavailable in a dataset.

Common representations: NaN, none,  NULL,  N/A, ""

We need to identify missing values before training an ML model.

### Step 2 – Real-World Analogy; 

Suppose a house record is: Area = 1500 sqft,  Bedrooms = 3,  Bathrooms = NaN,  Price = 85 lakhs

The bathroom information is missing.

A machine-learning model may not work correctly if missing values aren't handled properly.

### Step 1 – Understand the Concept
Own Words; Invalid values are values that don't make sense for the particular column.

Definition; An invalid value violates the expected range, meaning, format, or business rule of a variable.

### Step 2 – Real-World Analogy

A house cannot have:

Area = -1000 sqft
Bedrooms = -3
Price = -50 lakhs
Location Score = 25/10

These values are invalid.

### Step 3 – Implement in Python

First define numerical columns:

```python
numeric_columns = [
    "Area_sqft",
    "Bedrooms",
    "Bathrooms",
    "Age_years",
    "Location_Score",
    "Parking_Spaces",
    "Price_lakhs"
]
```

```python
null_like_values = [
    "",
    " ",
    "NULL",
    "null",
    "Null",
    "None",
    "none",
    "N/A",
    "n/a",
    "NA",
    "na",
    "?",
    "-",
    "Unknown",
    "unknown"
]
```

```python
data_clean = data.copy()
```

```python
data_clean = data_clean.replace(null_like_values,np.nan)
```

```python
data_clean.isnull().sum()
```

```python
data.dtypes
```

```python
data_clean["Area_sqft"] = data_clean["Area_sqft"].fillna(
    data_clean["Area_sqft"].median()
)
```

```python
data_clean["Bedrooms"] = data_clean["Bedrooms"].fillna(data_clean["Bedrooms"].median())

print(data_clean.head(50))
```

```python
data_clean["Age_years"] = data_clean["Age_years"].fillna(data_clean["Age_years"].median())
data_clean["Parking_Spaces"] = data_clean["Parking_Spaces"].fillna(data_clean["Parking_Spaces"].median())

print(data_clean.head(50))
```

## Handle Negative values to positive

```python
data_clean["Location_Score"] = pd.to_numeric(data_clean["Location_Score"], errors="coerce")
```

```python
## _ values comverting to nan and median aslo
data_clean.loc[data_clean["Location_Score"] <= 0, "Location_Score"] = np.nan
data_clean.loc[data_clean["Bedrooms"] <= 0, "Bedrooms"] = np.nan

data_clean["Location_Score"] = data_clean["Location_Score"].fillna(data_clean["Location_Score"].median())
```

```python
# converting _ values and 0 values into NAN
data_clean.loc[data_clean["Price_lakhs"] <=0, "Price_lakhs"] = np.nan
data_clean.loc[data_clean["Parking_Spaces"] <=0, "Parking_Spaces"] = np.nan
data_clean.loc[data_clean["Bathrooms"] <=0, "Bathrooms"] = np.nan
```

```python
## multiple columns converting NAN to median
multiple_col = ["Area_sqft", "Bedrooms", "Bathrooms", "Parking_Spaces", "Price_lakhs"]

for col in multiple_col:
    data_clean[col] = data_clean[col].fillna(data_clean[col].median())
```
